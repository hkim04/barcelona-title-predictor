```python
import pandas as pd

df_clean = all_matches[
    ["season", "Date", "HomeTeam", "AwayTeam",
     "FTHG", "FTAG", "FTR"]
].copy()

df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    dayfirst=True,
    errors="coerce"
)

print(df_clean.dtypes)
print("Invalid dates:", df_clean["Date"].isna().sum())
```

```python
import numpy as np

df_clean["ExpectedResult"] = np.select(
    [
        df_clean["FTHG"] > df_clean["FTAG"],
        df_clean["FTHG"] < df_clean["FTAG"]
    ],
    ["H", "A"],
    default="D"
)

mismatches = df_clean[
    df_clean["FTR"] != df_clean["ExpectedResult"]
]

print("Result mismatches:", len(mismatches))
```
